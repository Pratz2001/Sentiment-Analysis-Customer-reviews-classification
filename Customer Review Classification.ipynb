```python
import re
import string
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import SnowballStemmer
from nltk.corpus import wordnet
from nltk import pos_tag
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
def clean_text(text:str):
  text = str(text)
  text = text.lower()
  text = text.strip()
  text = re.sub(' \d+', ' ', text)
  text = re.compile('<.*?>').sub('', text)
  text = re.compile('[%s]' % re.escape(string.punctuation)).sub(' ', text)
  text = re.sub('\s+', ' ', text)
  
  text = text.strip()

  #removing stopwords
  filtered_sentence =[]
  stop_words = set(stopwords.words('english'))
  words = word_tokenize(text)
  for w in words:
    if w not in stop_words:
      filtered_sentence.append(w)
  text = " ".join(filtered_sentence)

  #stemming
  text = str(text)
  stemmed_sentence =[]
  snow = SnowballStemmer('english')
  for w in words:
    stemmed_sentence.append(snow.stem(w))
  text = " ".join(stemmed_sentence)

  return text
  
```

```python
data['text'] = data['reviews.text'].apply(clean_text)
```

```python
data = pd.read_csv('cleaned_data.csv')
data.head()
```

```python
data.dropna(inplace=True)
data[data['text'].isnull()]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
```

```python
vectorizer = TfidfVectorizer()
vectorizer.fit(data['text'])
features = vectorizer.transform(data['text'])
features.toarray()
```

```python
tf_idf = pd.DataFrame(features.toarray(), columns=vectorizer.get_feature_names_out())
tf_idf.head()
```

```python
Xtrain,Xtest,ytrain,ytest = train_test_split(tf_idf,data['sentiment_score'],test_size=0.2,random_state=42)
print(f'train set shape\t{Xtrain.shape}\nTest set shape\t{Xtest.shape}')
```

```python
yy = pd.DataFrame(ytrain)
train_data = pd.concat([Xtrain, yy],axis=1)
train_data.head()
```

```python
Xtrain = train_data.iloc[:,:-1]
ytrain = train_data['sentiment_score']
```

```python
ytrain
```

```python
from sklearn.metrics import confusion_matrix
from sklearn.metrics import roc_auc_score
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.naive_bayes import MultinomialNB
from xgboost import XGBClassifier
```

```python
def modeling(Model,Xtrain=Xtrain,Xtest=Xtest):
  model = Model
  model.fit(Xtrain,ytrain)
  global ypred
  ypred = model.predict(Xtest)
  print(classification_report(ytest,ypred))
  print ('AUC ',roc_auc_score(ytest, ypred))
  cm = confusion_matrix(ytest, ypred)
  print(cm)
```

```python
modeling(MultinomialNB())
```

```python
modeling(XGBClassifier())
```
